**Data subsetting**

So, the first thing we need to do, is to subset the dataframe to include columns and rows that are revelant. This is the time when we introduce data subsetting. If we want look at observations of China's economy from 1949 up until 2020, we will need to subset 1) observations of China and 2) observations from 1949 to 2020.

This gives you all rows about China:

````python
df.loc[df.country_name == "China"]
````

This gives you all rows between year 1949 and 2020:

````python
df.loc[(df.year >= 1949) & (df.year<=2020)]
````

And this gives you the intersection of both:

````python
df.loc[(df.country_name == "China") & (df.year >= 2000) & (df.year <= 2020)]
````

NOTE: You CANNOT use the base Python keywords such as "and", "or". They won't work. You need to use logic denotations, such as &, and |.

Now, we have successfully subsetted the rows that we want. But remember, there are over 4 thousands columns. We certainly do not want to deal with that many columns. How do we subset columns? The answer is also df.loc. Let's look at subsetting columns only first.

````python
df.loc[:,"country_name"]
````

You just got one column. If you want multiple columns, you need to pass in a list of strings of column names.

````python
df.loc[:,["country_name", "year", "e_gdp", "e_gdppc"]]
````

The syntax of .loc works as follows:

* df.loc[row-wise subsetting criteria, column-wise subsetting criteria]

* “:” means SELECT ALL

* df.loc[:, ['country_name', 'year', 'e_gdp', 'e_gdppc']] means, “select all rows, and subset this list of columns”

Q: Wait, but when we subset row (e.g., df.loc[df.country_name == "China"]), we used no “:”.

A: The above expression is an abbreviated version of df.loc[df.country_name == "China", :].

Pandas allow us to omit the subsetting conditions regarding columns if subset by row only.

However, if we subset by column only, we cannot omit the subsetting condition regarding rows.

Q: What are the inputs of the selection criteria?

A: Column/ row names, put in a list. E.g., ['country_name', 'year', 'e_gdp', 'e_gdppc']

Or, A list/ array/ column of booleans (Trae/ False). True means the row/ column should be included, False means the the row/ column should be excluded.
Note: It should have the same length as the dimension

Now, let's piece everything together to subset the dataframe and get:

1. Observations that are related to China
2. Observations that are between 1949 and 2020
3. Variables that are about economy, i.e., e_gdp and e_gdppc

Observation is about rows, so the first two criteria go in the first slot. Variables are about columns, so the last one goes into the second slot.

````python
df.loc[(df.year<=2020) & (df.year>=1949) & (df.country_name == "China"), ["country_name", "year", "e_gdp", "e_gdppc"]]
````

If you find it very hard to write and read, you can also create a manual filter before doing the subsetting. This is my personal favourite way to subset. Note you cannot use the builtin keyword "filter". That will override the builtin function.

````python
filt = (df.year<=2020) & (df.year>=1949) & (df.country_name == "China")
df.loc[filt, ["country_name", "year", "e_gdp", "e_gdppc"]]
````

Alternatively, Prof. Chen also demonstrated another way, which is to break the long line of code into several lines using the slash.

````python
df_china_econ = df.loc[(df.country_name == "China") \
       & (df.year >= 1949) \
       & (df.year <= 2020) \
       , ["country_name", "year", "e_gdp", "e_gdppc"]]
````

Okay! That is all we need to know about subsetting. There are some good practices that Prof. Chen talked about.

1. Subset step by step. Don’t do all operations in one command.
2. Use comments to document the subsetting operation
3. Save the subset of interest for your data analysis
4. Avoid repeatedly clean and wrangle the data
5. Use informative file names
6. Keep the original/ raw data, even if you don’t think you need them any more.
E.g., if you conduct a study on China and do the example operations for subset, you are strongly recommended to keep the global data

**Data Transforming**

After we obtained this China Economy dataframe, we might want to transform this dataframe so that it is easier for us to perform some analysis.

First, we want to transform a numeric variable to another numeric variable. Let's say the GDP numbers are too large and we want to move the decimal point forward by three, i.e., dividing GDP by 1000. How do we do that?

We simply create another column using exisitng columns, like this:

````python
df_china_econ["gdp_1000"] = df_china_econ["e_gdp"]/1000
````

The above is the same as:

````python
df_china_econ["gdp_1000"] = df_china_econ.e_gdp/1000
````

That is to say, when you refer to an existing column inside a dataframe, you can use the attribute-like way. If you want to create a column, the canonical square bracket is the only way to go.

Next, sometimes economists work with log data to capture some pattern. What if we want to transform the GDP value to a log scale? To do that, we need another library called numpy.

````python
import numpy as np
df_china_econ['gdp_log'] = np.log(df_china_econ['e_gdp'])
````

Now let's look at what the dataframe is like:

````python
df_china_econ
````

Next! Now we want to transform a numeric variable to binary variable. We want to add another column that marks if the year is under Xi's administration.

````python
df_china_econ["admin_xi"] = (df_china_econ["year"] <= 2023) & (df_china_econ["year"] >= 2013)
````

````python
df_china_econ
````

That does not look pretty nice. I want to have all Chinese presidents in this column. I need to remove the admin_xi column first.

````python
df_china_econ.drop(columns = ['admin_xi'], inplace = True)
````

You can also do it using assignment:



```

```

And you can drop multiple columns at once, as long as you pass in a list for the columns argument.

```

```




````python
df_china_econ
````

Next, I need to create a new column where each year has a name of the presidents, i.e., Mao, Hua-Hu-Zhao, Jiang, Hu, Xi.

* Mao: 1949-1976
* Hua-Hu-Zhao: 1976-1989
* Jiang: 1989-2002
* Hu: 2002-2013
* Xi: 2013-2023


````python
df_china_econ['admin'] = None # Initiate a column named admin with None type
df_china_econ.loc[(df_china_econ.year >= 1949) & (df_china_econ.year < 1976), 'admin'] = "Mao"
df_china_econ.loc[(df_china_econ.year >= 1976) & (df_china_econ.year < 1989), 'admin'] = "Hua-Hu-Zhao"
df_china_econ.loc[(df_china_econ.year >= 1989) & (df_china_econ.year < 2002), 'admin'] = "Jiang"
df_china_econ.loc[(df_china_econ.year >= 2002) & (df_china_econ.year < 2013), 'admin'] = "Hu"
df_china_econ.loc[(df_china_econ.year >= 2013) & (df_china_econ.year <= 2023), 'admin'] = "Xi"
````

````python
df_china_econ
````

**Data Aggregating**

Now everything looks pretty good. I need some statistics to see which years are the best and which ones are the worst in terms of economic performance. So it is a good time for us to introduce "aggregate".

If you are familiar with R, this notion could be very straightforward to you. Aggregation in Python is pretty much like summarise() in R, which more limited functionalities.

````python
df_china_econ["gdp_1000"].agg(['mean', 'std', 'min', 'max'])
````

If you want to group by administration, you can add in this groupby method. It will give you these statistics for each value of the column grouped. So here, for each value inside the admin column, you get mean, std, min, and max.



````python
df_china_econ.groupby('admin')['gdp_1000'].agg(['mean', 'std', 'min', 'max'])
````

What if you want different statistics for different columns? Can you have different aggragation for each of the columns? Of course. See below:

````python
df_china_econ.agg({'gdp_1000': ['mean', 'min', 'max'], 'e_gdppc': ['mean', 'min', 'max', 'std']})
````

````python
df_china_econ.groupby('admin').agg({'gdp_1000': ['mean', 'min', 'max'], 'e_gdppc': ['mean', 'min', 'max', 'std']})
````

Dr Chen recommends this approach. It looks nicer.

````python
df_china_econ.groupby('admin').agg(
    gdp_mean = ('gdp_1000', 'mean'),
    gdp_min = ('gdp_1000', 'min'),
    gdp_max = ('gdp_1000', 'max'),
    gdp_std = ('e_gdppc', 'std'),
    gdppc_mean = ('e_gdppc', 'mean'),
    gdppc_min = ('e_gdppc', 'min'),
    gdppc_max = ('e_gdppc', 'max'),
    gdppc_sd = ('e_gdppc', 'std')
).reset_index()
````

The reason why we reset the index is that we do not want those values inside the admin column to be the index. Python automatically set the row index to be the values that are grouped. By reset_index, we create another set of integers from 0 to be our index for each row.

````python
sum_by_admin = df_china_econ.groupby('admin').agg(
    gdp_mean = ('gdp_1000', 'mean'),
    gdp_min = ('gdp_1000', 'min'),
    gdp_max = ('gdp_1000', 'max'),
    gdp_std = ('gdp_1000', 'std'),

    gdppc_mean = ('e_gdppc', 'mean'),
    gdppc_min = ('e_gdppc', 'min'),
    gdppc_max = ('e_gdppc', 'max'),
    gdppc_std = ('e_gdppc', 'std'),
).reset_index()
````

````python
sum_by_admin
````

Let's save that as sum_by_admin. Another function we want to let you know is idxmax() and idxmin(). They give you the index of first occurrence of maximum over requested axis. NA/null values are excluded. For example, if you want to know when China performed the worst or the best under the administration of each leaders, you can do the following:

df_china_econ.groupby('admin').gdp_1000.idxmin()

**Data Renaming**

New task: We want to compare the economy of China and Japan from 1949 to 2019.

What we need: Create a new subset named df_japan_econ. In the same way, we filter out the country name of Japan, with year of 1949 to 2019, and only include columns that are relevant.

````python
df_japan_econ = df.loc[(df.country_name == "Japan") & (df.year >= 1949) & \
                        (df.year <= 2019), \
                         ['country_name', 'year', 'e_gdp', 'e_gdppc']]
````

````python
df_japan_econ
````

Now because we have two different countries, we might need to rename their column names so that there is no confusion when the data is joined together. How do you rename variables? Use dictionary! **You make the original column name the key of that dictionary, and the desired new name the value of that dictionary.** And use the rename method.

````python
{'e_gdp': 'gdp_china', 'e_gdppc': 'gdppc_china'}
````

````python
df_china_econ_h = df_china_econ[['year', 'e_gdp', 'e_gdppc']].\
  rename(columns = {'e_gdp': 'gdp_china', 'e_gdppc': 'gdppc_china'})
````

````python
df_china_econ_h
````

````python
df_japan_econ_h = df_china_econ[['year', 'e_gdp', 'e_gdppc']].\
  rename(columns = {'e_gdp': 'gdp_japan', 'e_gdppc': 'gdppc_japan'})
````

````python
df_japan_econ_h
````

**Data Combining**

To combine datasets horizontally (i.e., placing columns of two datasets side by side), we use pd.merge().

````python
df_cn_jp = pd.merge(df_china_econ_h, df_japan_econ_h, on = 'year')
````

By contrast, to combine datasets vertically (i.e., placing data entries of dataset B at the bottom of dataset A), we use pd.concat().

````python
df_china_econ_v = df_china_econ.loc[:, ['country_name', 'year', 'e_gdp', 'e_gdppc']]
df_japan_econ_v = df_japan_econ.loc[:, ['country_name', 'year', 'e_gdp', 'e_gdppc']]
````

````python
pd.concat([df_china_econ_v, df_japan_econ_v])
````
