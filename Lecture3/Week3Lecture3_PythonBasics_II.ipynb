### 3.2 Control Flows

Control flow (or flow of control) is the order in which individual statements, instructions, or function calls of an imperative program are executed or evaluated.

For now, all of our programs are running linearly. That actually is a type of control flow, which is sequential.

There are other types of control flows, including conditions and loops.

Let's look at conditionals first. Let's check if the first element of gdp_mexico_list is greater to 100000. If so, print YES.

#### 3.2.1 Conditionals

A conditional statement consists of the following elements.

```
if condition 1:
  (something you want to do if condition 1 is met)
elif condition 2:
  (something you want to do if condition 2 is met)
else:
  (something you want to do if none of the previous conditions is met)
```

The elif and else parts are optional.

````python
if gdp_mexico_list[0] > 100000:
  print("YES, it is > 100000")
````

Note that this is one important feature of Python. It uses indentation to indicate the relationships within structures. The indentation here means that the `print` command is a part of the `if `statement.

````python
if gdp_mexico_list[0] > 100000:
  print("YES, it is > 100000")
else:
  print("NO, it is not > 100000")
````

````python
if gdp_mexico_list[0] > 100000:
  print("YES, it is > 100000")
elif gdp_mexico_list[0] > 150000:
  print("YES, it is > 150000")
else:
  print("YES, it is not > 100000")
````

Another feature of these conditionals is that they will stop after the first occurance of the condition that is met. All following ones will be skipped. So the order matters!

````python
if gdp_mexico_list[0] > 150000:
  print("YES, it is > 150000")
elif gdp_mexico_list[0] > 100000:
  print("YES, it is > 100000")
else:
  print("YES, it is not > 100000")
````

We can also check something else, if it's an integer, and if it's bigger than something, using some logical operators.



````python
to_check = 2.1
````

````python
# Want: Check if it is an integer and if >=3
if isinstance(to_check, int) and to_check >= 3:
  print("INT & >=3")
elif isinstance(to_check, int) and to_check < 3:
  print("INT & <3")
elif (not isinstance(to_check, int)) and to_check >= 3:
  print("not INT & >=3")
elif (not isinstance(to_check, int)) and to_check < 3:
  print("not INT & <3")
else:
  print("nothing happens")
````

The 'else' part is just to capture all other situations that were not specified in previous if statements.

### 3.3 Functions

What is a function like?

*   A function is a block of code which only runs when it is called.
*   You can pass data, known as parameters, into a function.
*   A function can return data as a result.

What is a function for?

* A function bundles a set of instructions that you can use repeatedly.



A user-defined function has its name and body, oftentime it also takes some arguments. In the following function, the name is "calculate_mean_of_list". It takes one **local** argument of input_list, and print out the mean_of_list.

The concept of local object is very important. In the following function, we have an input argument of input_list. When we call that function, this input list might be a list of [0,1,2,3,4]. This list is assigned to that name of input_list only within that function. If there is no such object in the global environemtn and you try to call that in the global environment, there will be errors. After this function is executed, this temporary mapping between the list and that name will be thrown away.

The same is true for other names within that function. In our case, that includes things like sum_os_list, number_of_element, mean_of_list and so on. They are only useful to this specific function.

````python
def calcualte_mean_of_list(input_list):
  # Two steps for calculating the means
  # First: get the sum
  # Second: devide the sum by the nubmer of elements
  sum_of_list = sum(input_list)
  number_of_element = len(input_list)

  mean_of_list = sum_of_list / number_of_element
  # Print results
  print(mean_of_list)
````

A function will be executed when you call that function.

````python
calcualte_mean_of_list(gdp_mexico_list)
````

Why it is a Nontype? Ah, in out user-defined function of "calculate_mean_of_list", we did not return anything. We only printed out the result. But print is a function that returns nothing. When there is nothing to return, we actually saved the "nothing" to an object called result. So when we call that result, there is nothing. And it is a NoneType.

If we modify our function, adding something to be returned, we will solve this problem.



````python
def calcualte_mean_of_list(input_list):
  # Two steps for calculating the means
  # First: get the sum
  # Second: devide the sum by the nubmer of elements
  sum_of_list = sum(input_list)
  number_of_element = len(input_list)
  mean_of_list = sum_of_list / number_of_element
  return(mean_of_list)
````
