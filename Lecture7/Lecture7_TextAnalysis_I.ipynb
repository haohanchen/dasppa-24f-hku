**Objective of Text Analysis**

We use machine to augment our capacity to READ text, including:

- Summarize the text

- Look for specific contents of interest

- Search for text related to the current one

**Unit of analysis**

Anslysis of texts can be done at different levels, including at the corpus level (i.e., the entire collection of documents), in which case we are interested in:

* What contents/ concepts are in this corpus? How frequent does each one appear?

* How frequently does the corpus discuss certain topic/ concept of your choice?

Or at the document level:

* What contents/ concepts are in this document? How frequent does each one appear?

* How frequently does the document discuss certain topic/ concept of your choice?

* Which documents is most similar to the document of interest?

Or only part of document (e.g., paragraph, sentence, word, phrase)

* Which paragraphs/ sentences/ words/ phrases are similar to the one of interest?


**Modern Approach**

Modern approach of text anslysis is the kind of methods before the recent boom of Large Language Models (i.e., LLM, e.g., ChatGPT, or the post-modern approach)
Despite the recent developments, modern approach is essential for forming basic understanding of text analysis and will be useful for years even with the new methods.

Some features for the modern approach:

* Computationally efficient: You do not need 100 highly powerful and expensive GPUs to train your model, like OpenAI and Deepseek does. You can perform modern text analysis even using your laptop.

* Good for data safety: Everything that Python processes remains in the memory and the machine that you use to process the text data. Nothing will be transferred to any third party. Unlike commercial large langauge models that might use your input as their training data. Also, the models that we employ are open-source. This means everyone can see their original codes and review the data security of these models.

* Inexpensive: Basically free, if you run it with Google Colab, which provides free computational resrouces.

### 1. Extract Linguistic Features from Text

Extracting linguistic feature (“linguistic annotation”) is an essential pre-step in the modern text analysis approach. Techniques that we are primarily interested in include:
* Tokenization: Segmenting text into words, punctuations marks etc. Usually for English, the boundaries are drawn by a space between two tokens. SpaCy does this automatically for us.

* Stemming/Lemmatization: Assigning the base forms of words. For example, the lemma of “was” is “be”, and the lemma of “rats” is “rat”. In other words, it changes words that are inflected for gramatical reasons back to their "basic" forms.

* Part-of-speech tagging: Assigning word types to these tokens, such as "work" is a verb and "China" is a noun.

* Stopword removal: Remove words that occur commonly across documents. They are typically not substantively meaningful. Examples: “the” “a” “to”, "of".


To perform this, we use SpaCy, a tool for extracting linguistic and substantive features with only a few lines of code.

https://spacy.io/
It is very friendly for beginners, free, easy to use, nultifunctional and multilingual.

````python
import spacy
nlp = spacy.load("en_core_web_sm")
````

**Getting Linguistic features using List Comprehension**

These are the commands that we use to get these features. Remember from the beginning of the semester where we introduced "list comprehension"? Let's go over it once again.

Originally, if we want to create a list, we would:

````python
# I want to creata a new list that consists of fruit names that contains English letter "a".
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newlist = []

for x in fruits:
  if "a" in x:
    newlist.append(x)

print(newlist)
````

But that seems to be very long and excessive for many programmers. With list comprehension, we can significantly shorten that into:

````python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]

newlist = [x for x in fruits if "a" in x]

print(newlist)
````

With this, you are practically running a for loop within that newly created list. In the same way, we get get the linguistic features relatively quickly with list comprehension:

Get raw tokens:

	`[word for word in doc]`

I.e., Include the word, if the word exists in the document.

	`[word.text for word in doc]`

Get low-cased tokens

	`[word.lower_ for word in doc]`
  
I.e., include the lowercase of the word, if the word exists in the document.

Get lemmatized tokens

	`[word.lemma_ for word in doc]`

I.e., include the lemmatized word, if the word exists in the document.

Get part-of-speech tags of tokens

	`[word.pos_ for word in doc]`

Get stopword indicators for tokens

	`[word.is_stop for word in doc]`

To make a dataframe that contains everything

```
pd.DataFrame(
  {
      'token': [word for word in doc],
      'lower': [word.lower_ for word in doc],
      'lemma': [word.lemma_ for word in doc],
      'pos': [word.pos_ for word in doc],
      'stopword': [word.is_stop for word in doc]
  }
)
```
The interpretation of the above is that: We are creating a dictionary of lists, where the list names are the key and the lists themselves are the values. And then we use this dictionary to create a dataframe, which takes the keys as the column names, and make the lists the columns themselves. Let's apply this to one document, which is the first question in this dataset.

````python
# Tokenize the first question using nlp().
doc = nlp(df.loc[0, "question"])
````

````python
# Get all linguistic features and print out the dataframe.
ling_features = pd.DataFrame(
  {
      'token': [word for word in doc],
      'lower': [word.lower_ for word in doc],
      'lemma': [word.lemma_ for word in doc],
      'pos': [word.pos_ for word in doc],
      'stopword': [word.is_stop for word in doc]
  }
)
ling_features
````

### 3. Functionizing and batch processing

**Creating a Function of Extracting Linguistic and Entity Features**

What if you want to make both processes more automatic? In that way, you can just specify the texts you need to analyze, then the rest of the job is for Python. It is actually not difficult. All you need is to automate the tokenization process.

````python
def text_analysis(text_input):
  doc = nlp(text_input)
  extract_ling = pd.DataFrame(
    {
        "token": [word.text for word in doc],
        "lower": [word.lower_ for word in doc],
        "lemma": [word.lemma_ for word in doc],
        "pos": [word.pos_ for word in doc],
        "stopword": [word.is_stop for word in doc],
    })

  extract_ne = pd.DataFrame(
    {
        'ne': [ne.text for ne in doc.ents],
        'ne_lemma': [ne.lemma_ for ne in doc.ents],
        'ne_label': [ne.label_ for ne in doc.ents],
        'ne_start': [ne.start for ne in doc.ents],
        'ne_end': [ne.end for ne in doc.ents],
        'ne_start_char': [ne.start_char for ne in doc.ents],
        'ne_end_char': [ne.end_char for ne in doc.ents],
    })
  return extract_ling, extract_ne
````

IMPORTANT: What have we achieved? We have just turned unstructured data into structured data.

**Think about being economic and the memory management**

We created a function that first analyzes all features that we might need, then stores the features in dictionaries and corresponding dataframes. This is suitable for processing documents that are not exceedingly long, such as MoFA press conference for a single day ot week.If you would like to do this with 1000 documents, it is very likely that you will run out of memory. And the processing speed will ve very slow. See the next section for batch processing.

**Extracting Features from a Collection of Documents**

We have just performed all these analyses to one document. What shall we do are to scale it up for a collection of documents? For example: Extract features from all the questions asked in the MoFA conferences in 2023?


Using the function we created, we can:

1. Create a subset of the dataframe covering Q&A in 2023

2. Locate the “Question” column

3. Repeatedly perform the same text analysis to every question


````python
# Filter out the rows of 2023
df_2023 = df.loc[df.year == 2023, :].reset_index()

# Create two lists that we store the output dataframes
result_ling_ls = []
result_ne_ls = []

# Write a for loop to loop through all texts
for id, text in zip(df_2023.id, df_2023.question): # Remember our old friend zip()?
  r_ling, r_ne = text_analysis(text)

  # Store the id that will be used later for merging
  r_ling['id'] = id
  r_ne['id'] = id

  # Append each resulting dataframe to the list
  result_ling_ls.append(r_ling)
  result_ne_ls.append(r_ne)

  # Uncomment if you need to check the process
  #print(id)

# Finally, concatonate the two lists of dataframes into two huge dataframes
text_ling = pd.concat(result_ling_ls, ignore_index = True)
text_ne = pd.concat(result_ne_ls, ignore_index = True)
````
