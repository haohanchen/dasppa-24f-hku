## 1. Philosophy of Data Visualization

Here are two questions for you to think about:

* What is data visualization?
* Why do we (you) care about data visualization?

Data visualization is basically converting numbers into graphs. There is a saying that a picture is worth a thousand words, in our case, numbers. It is true. Humans are just biologically and mentally willing to and good at taking visual information! There are some very important use of data visualizations.

For you as a data analyst:

* Summarize data and results from analysis
* Discover from data and data analysis

For you as a writer and presenter:

* Showcase rigor and complexity
* Highlight insights
* Simplify interpretation

For you as a consumer:

* Get the point quickly
* Catch misinterpretation (intentional or not)
* Inquire the right information

**Therefore, good visualization should:**

* Showcase complexity, but not over-complicate
* Highlight important data and be faithful to the raw data
* Simplify unnecessary things, but don't overlook critical information.
* Perhaps very importantly: Good visualization is not the same as beautiful data visualization

### One numeric variable

**Histogram**

A histogram is a chart that plots the distribution of a numeric variable's values as a series of bars. Each bar typically covers a range of numeric values called a bin or class; a bar's height indicates the frequency of data points with a value within the corresponding bin.

```python
# With e_gdppc
sns.histplot(data = df_s, x = "e_gdppc")
```

```python
# With e_pelifeex
sns.histplot(data = df_s, x = "e_pelifeex")
```

**KDE Plot**

A kernel density estimate (KDE) plot is a method for visualizing the distribution of observations in a dataset, analogous to a histogram. KDE represents the data using a continuous probability density curve in one or more dimensions.

```python
sns.kdeplot(data = df_s, x = "e_gdppc")
```

```python
sns.kdeplot(data = df_s, x = "e_pelifeex")
```

**Distribution plot**

When you use distplot() with out specifying anything, the dafault output will be exactly a histogram..

```python
sns.displot(data = df_s, x = "e_gdppc")
```

```python
sns.displot(data = df_s, x = "e_gdppc", kind = "kde")
```

**Displot + KDEplot**

You can have two different types of visualizing distributions in the same plot. Just specify "ked = True".

```python
sns.displot(data = df_s, x = "e_gdppc", kde = True)
```

```python
sns.displot(data = df_s, x = "e_pelifeex", kde = True)
```

**Since distplot() is a unified command for both histogram and kde plot, I recommend memorizing this command for simplicity.**

**Customizations**

* The number of bins in a plot. This also changes the width of each bin. If you have 30 bins for a data range of 30, each bin will be exactly 1 in width.

```python
sns.displot(data = df_s, x = "e_pelifeex", bins = 20)
```

### Two Numeric Variables

* Scatterplot

```python
sns.scatterplot(data = df_s, x = "e_gdppc", y = "e_pelifeex")
```

**Scatterplot + Histogram (Jointplot)**

```python
sns.jointplot(data = df_s, x = "e_gdppc", y = "e_pelifeex")
```

**Jointplot + Regression line**

The association does look convincing with the fitted line from regression? **But this only serve to assist your pattern discovery. It should not replace rigorous machine/ statistical modeling.**

Also, if you find the two quantitative variables to have non-linear correlation, it is not a good idea to add a linear fitted line in your data visualization. Because that would violate the assumptions of linear regression.



```python
sns.jointplot(data = df_s, x = "e_gdppc", y = "e_pelifeex", kind = "reg")
```

### One categorical variable

Since we do not have any categorical variables yet, let's create one. How about grouping by different regions?

```python
df_s.e_regiongeo.value_counts()
```

```python
df_s.loc[:, 'region'] = None
```

```python
df_s.loc[df_s.e_regiongeo.isin((1, 2, 3, 4)), 'region'] = "Europe"
df_s.loc[df_s.e_regiongeo.isin((5, 6, 7, 8, 9)), 'region'] = "Africa"
df_s.loc[df_s.e_regiongeo.isin((10, 11, 12, 13, 14)), 'region'] = "Asia"
df_s.loc[df_s.e_regiongeo == 15, 'region'] = "Oceania"
df_s.loc[df_s.e_regiongeo.isin((16, 17, 18, 19)), 'region'] = "Americas"
```

```python
df_s.region.value_counts()
```

Tip: A good practice is that you should always tabulate categorical variables before conducting analysis.


```python
pd.crosstab(df_s.e_regiongeo, df_s.region)
```

```python
df_s.loc[:, 'income_level'] = None
df_s.loc[df_s.e_gdppc < 1.025, 'income_level'] = "Low"
df_s.loc[(df_s.e_gdppc >= 1.025) & (df_s.e_gdppc < 3.995), 'income_level'] = "Low-middle"
df_s.loc[(df_s.e_gdppc >= 3.995) & (df_s.e_gdppc < 12.375), 'income_level'] = "High-middle"
df_s.loc[df_s.e_gdppc >= 12.375, 'income_level'] = "High"
```

The plot we aim to replicate also includes income level. Let's make a categorical variable for that as well.

```python
df_s.income_level.value_counts()
```

**Countplot**

```python
sns.countplot(data = df_s, x = "income_level")
```

## Prettifying visualizations

Now that your plot gets more complicated, it’s time to talk about labeling and aesthetics…

The top priority is still to make **clean and informative** visualizations.

The tedious but essential thing: Make your figures readable, including:

* Add the title of the figure and labels of x and y axes
* Adjust figure sizes
* Choose a theme you like


```python
# Add the title of the figure and labels of x and y axes
sns.scatterplot(data = df_s, x = "e_gdppc", y = "e_pelifeex",
                size = "e_pop", sizes = (20, 2000), hue = "region",
                legend = False, alpha = 0.5).\
                set(
                    title = "Health and Wealth (2019)",
                    xlabel = "GDP per capita",
                    ylabel = "Life Expectancy"
                )
```

```python
# Adjust figure size
plt.figure(figsize = (10, 8))

sns.scatterplot(data = df_s, x = "e_gdppc", y = "e_pelifeex",
                size = "e_pop", sizes = (20, 2000), hue = "region",
                legend = False, alpha = 0.5).\
                set(
                    title = "Health and Wealth (2019)",
                    xlabel = "GDP per capita",
                    ylabel = "Life Expectancy"
                )
```

Add legends

It is a good practice to include the legends! However:
* Challenge: No quick and easy way to re-label the legends.
* We suggest you changing the variable names before ploting it.


```python
plt.figure(figsize = (10, 8))

sns.scatterplot(data = df_s, x = "e_gdppc", y = "e_pelifeex",
                size = "e_pop", sizes = (20, 2000), hue = "region",
                legend = True, alpha = 0.5).\
                set(
                    title = "Health and Wealth (2019)",
                    xlabel = "GDP per capita",
                    ylabel = "Life Expectancy"
                )
```

You can also set a theme that you like. The default is normally fine, but you can try out:

``
sns.set_style("white")
``

``
sns.set_style("whitegrid")
``

``
sns.set_style("darkgrid")
``

``
sns.set_style("dark")
``

Note: these commands have *global* impact. Once set, they impacts over every figure you plot afterwards. You typically put it at the start of a notebook.


```python
sns.set_style("darkgrid")

plt.figure(figsize = (10, 8))

sns.scatterplot(data = df_s, x = "e_gdppc", y = "e_pelifeex",
                size = "e_pop", sizes = (20, 2000), hue = "region",
                legend = True, alpha = 0.5).\
                set(
                    title = "Health and Wealth (2019)",
                    xlabel = "GDP per capita",
                    ylabel = "Life Expectancy"
                )
```
