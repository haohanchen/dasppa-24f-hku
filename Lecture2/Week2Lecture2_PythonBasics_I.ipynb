## 1. Motivations for programming and using Python.

**Why programming?**

Why do we program to do data science? Why can't we simply use commercial tools such as Microsoft Excel, Apple Numbers, SPSS, or other similar softwares? These tools will allow us to do many things without programming. We just open up a dataset and use those functions to do the math. It is that simply and easy.

However, programming does have at least five advantages over these tools. 

- First, **scalability**. That is programming in Python can let you deal with very larg data. If your PC or laptop isn't powerful enough, opening up a spreadsheet in Excel can easily freeze the whole machine. Programming wouldn't. It just takes some more time, but normally it will get things done.

- Second, **percision**. Interpreters of programming languages such as Python and R are very stupid in the sense that they only understand things you tell them that are consistent with the pre-defined syntax. This might sounds bad, but it is also very percise. You let the machine to do exactly what you tell it to.

- Then, **automation and replication**. Given a piece of code, whether it's you or other people, no matter it's today or next Thursday or next year, it will give us the same thing. If you published a paper and attached a piece of code, people can easily replicate your study and be more confident about your research.

- **More tool to come:** Many programming langauges have very big communities that are constantly developing and updating tools for us to use. You will get more tools and functions available in the future. Instead of paying more money to Microsoft, you can use these new tools for free.

- Finally, **customization of workflow.** These commercial tools only have fixed functions. You will need to adapt to what the developers want you to do. Programming languages set very few limits. Have you heard of Excel VBA, which is a programming language running on MS Office? Even Microsoft knows that their Excel is not omnipotent and they created a programming langauge for it, which is now used by many financial and business analysts.

Will Generative AI replace programmers? Do we still need programming skills? We still need it and AI will not replace, at least for now. The prompts you give to these AI models are human languages. They are not percise. Programmers can translate human language into specific commands, which is something AI is not fully competent at right now.

**Comments: Use the # sign**

Comments are easy way for you to note the purpose of codes (or in fact anything you want) along side with your codes. Comments will not be executed by Python. They can be either placed above or below codes:

```python
# Multiply 3 by 2
print(3 * 2)
```

...or after codes.

```python
print(3 * 2) # Multiply 3 by 2
```

But anything that comes after the # sign will become comments and therefore, not executed. Also, please do not write texts directly into any code snippet. It is very unprofessional:

```python
Multiply 3 by 2
```

**Variable**

Variable is something Python stores in its memory space. You can reuse them until they are changed to something else. To define a variable, you use the equal sign. This binds the name of the variable (e.g., `test_var` to the thing that follows the equal sign, e.g., the result of four plus five).

```python
# Define a variable named test_var
# test_var takes the value of 4 + 5
test_var = 4 + 5
# The above line of code save test_var in the memory
# To see its value, you need to print it
print(test_var)
```

Compare that with:

```python
print("test_var")
```

Because now you are specifically asking Python to print these letters together, not the variable you defined (without the quotation marks), you get "test_var", not "9".

Also, pay attention to the variables' nameing rules. **No space in the middle.**  **No `.` in the middle (This means something else in Python!).** And **no special characters.**

We recommend that names of variables should be:
1. Meaningful English words
2. Plus numbers
3. Spearated by underscores

```python
# Create a variable indicating the inflation rate.
inflation_rate = 4
# This is not a good practice:
r = 4
# Not informative!
```

**Manipulating Variables**

```python
my_var = 3
print(my_var)
```

```python
my_var = 100
print(my_var)
```

What do the following codes generate? Take a guess!

```python
my_var = my_var * 3
print(my_var) 
```

Assuming that you have executed the codes sequentially: Because you have updated the my_var variable to 100 in the second code snippet. Binding that variable name to 100 multiplied by 3 will give you 300, instead of 9, or 100.

**Multiple Variables**

```python
num_years = 4
days_per_year = 365
hours_per_day = 24
mins_per_hour = 60
secs_per_min = 60
```

Multiple variables are just what they represent. So the following command should give you the result of all five numbers combined. 

```python
total_secs = secs_per_min * mins_per_hour * hours_per_day * days_per_year * num_years
```

```python
print(total_secs)
```
