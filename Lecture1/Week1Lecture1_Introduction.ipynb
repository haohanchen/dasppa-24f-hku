# **2. Course Plan**

**Why data science and POLI3148?**

Currently, data science is extraordinary. Think about the big data industry and AI. In the future, data science will become ordinary, even compulsory. Think about typists in the 1990s. There were people who specifically expertized in typing in big companies. But now everybody can do that. Data science will be this kind of job very soon. Right now there are people who expertise in coding and data science. In 20 years, data scientists will be a skill available to everybody. You should learn a skill that will become ordinary in a couple of dacades to become an relatively early adoptor, or even an industry leader.

**Debunking two myths about data science**

Myth 1: Social sciences and humanity trainees/ professionals are losing jobs to, technicians (computer scientists, data scientists…), AI

* Yes and no

* Yes: mechanical and repetitive human jobs are being replaced. These are also areas that current AIs are very capable of.

* No: Deep understanding of the society matters. It takes a lot of humanities and social knowledge and experience to deeply understand social issues. People will be the final decision-maker in terms of implementing a policy, or design agenda in politics, at least for now.

Myth 2: You need to be good at maths/ technologies to learn data science.

* Yes and no

* Yes: learning the “why” needs background knowledge. For example, it takes a lot math knowledge to figure out why certain machine learning models behave like they do, or to know what under the hood of a text analysis tool. However, given the time and our background, and for the purpose of this course, we do not care so much about it.

* No: learning the “how” is getting easier -- consider it learning a language. You learn a language through practice and the use of it. We do it with Python in pretty much the same way. This course offers you opportunities just like that.

**Data Wrangling**

* Load, clean, and transform data for analysis

* In other words, cleaning messy data

The data you are working with might be very very messy. All in different formats, or missing of values, etc. Maybe also many human-introduced errors, or you need to combine different sources of data. Data wrangling deals with that. (This is also the most time-comsuming part of data science)

**Data Visualization**

Basically to turn numbers to graphics. You may be familiar with different kinds of visualization, such as scatterplot, bar charts, histograms. Python has a very user friendly visualization module to help you with that. People often say that a good plot says a thousand words. Data visualization is an important part of data science for you to communicate your findings to potential audience. We will cover some basics for you to get started and a path for you to go very advanced.

**Text Analysis**

The technique to use computational methods to analyze texts and documents. For example, you have the speech text of US Congress in the past century. You may want to understand the US-Sino relations. Text analysis can extract useful information about the texts. We have two sessions on text analysis. We will talk about things like tokenization, lemmatization, word frequency and word cloud, sentiment analysis, etc.

**Machine Learning**

It is a very broadly defined term. There are different models in thie realm. You may want to use them predict where do conflicts occur, or under which condition does a country have conflicts with another country?

Some more conventional models, such as regressions, others such as decision tree, and basics of neural networks will also be covered.

**Datasets/Cases we use**

To develop your “computational thinking” capacity, to help you understand which computational methods should be applied to political data, and to give meaningful interpretation to outputs of computational methods, we have the following datasets and cases.

* V-dem.net, A rich cross-national datasets with economic and political indicators in the past hundreds of years.

* ucdp.uu.se, Tracking global conflict data with geo-reference data. Geo-coded data  of organized violence (phenomena of lethal violence occurring at a given time and place).

* Chinese Foreign Ministry: Text transcripts of press conferences of China’s Ministry of Foreign Affairs. (Mochtak, Michal, and Richard Q. Turcsanyi. We answer the question with a Biographical dataset of Chinese officials and a machine learning approach. Paper and data available at: https://doi.org/10.1073/pnas.230514312 and https://chinadatalab.ucsd.edu/elites/

You are recommended to take a look and get a good idea about what their data structures are like before we actually use them.
